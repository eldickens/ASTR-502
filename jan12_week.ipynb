```python
from isochrones.mist import MIST_Isochrone
import matplotlib.pyplot as plt
mist = MIST_Isochrone()
```

```python
# compare two different isochrones : 1 Gyr and 5 Gyr
iso_1gyr = mist.isochrone(9.0, 0.0)  
iso_5gyr = mist.isochrone(9.7, 0.0)

plt.figure(figsize=(5,6))

# G = Gaia’s “broad optical band” (total brightness of the star)
# K = near-infrared filter (~2.2 μm wavelength)
# x = G - K  → "temperature"
# y = G      → "luminosity"

plt.plot(
    iso_1gyr['G_mag'] - iso_1gyr['K_mag'],
    iso_1gyr['G_mag'],
    label='1 Gyr'
)

plt.plot(
    iso_5gyr['G_mag'] - iso_5gyr['K_mag'],
    iso_5gyr['G_mag'],
    label='5 Gyr'
)

plt.gca().invert_yaxis()
plt.xlabel('G − K')
plt.ylabel('G')
plt.title('MIST Isochrones: 1 Gyr vs 5 Gyr')
plt.legend()
plt.show()
```

```python
# compare two different isochrones : [Fe/H] = 0.0 and [Fe/H] = -0.5 
iso_feh0  = mist.isochrone(9.3,  0.0)
iso_fehm5 = mist.isochrone(9.3, -0.5)

plt.figure(figsize=(5,6))

plt.plot(
    iso_feh0['G_mag'] - iso_feh0['K_mag'],
    iso_feh0['G_mag'],
    label='[Fe/H] = 0.0'
)

plt.plot(
    iso_fehm5['G_mag'] - iso_fehm5['K_mag'],
    iso_fehm5['G_mag'],
    label='[Fe/H] = -0.5'
)

plt.gca().invert_yaxis()
plt.xlabel('G − K')
plt.ylabel('G')
plt.title('Metallicity Effect at Fixed Age')
plt.legend()
plt.show()
```
