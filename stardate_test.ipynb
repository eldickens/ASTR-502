```python
import numpy as np
import stardate as sd
from stardate.lhf import age_model
```

```python
# solar test without rotation
iso_params_sun = {
    "teff": (5772, 5),         
    "logg": (4.438, 0.02), 
    "feh": (0.0, 0.01),        
    "parallax": (1000.0, 0.1),   
    "maxAV": 0.01
}

star = sd.Star(iso_params_sun)
star.fit()
age, errp, errm, samples = star.age_results()

print(f"\nSolar Age = {age:.2f} +{errp:.2f} -{errm:.2f} Gyr")
```

```python
# solar test with rotation
iso_params_sun = {
    "teff": (5772, 5),        
    "logg": (4.438, 0.02),    
    "feh": (0.0, 0.01),     
    "parallax": (1000.0, 0.1),  
    "maxAV": 0.01
}

# solar rotation period
sun_prot = 25.4
sun_prot_err = 0.5

star = sd.Star(iso_params_sun, prot=sun_prot, prot_err=sun_prot_err)
star.fit()
age, errp, errm, samples = star.age_results()

print(f"\nSolar Age = {age:.2f} +{errp:.2f} -{errm:.2f} Gyr")
```

```python
stars = [
    {
        "name": "AP Col",
        "params": {
            "teff": (3250, 100),
            "logg": (4.8, 0.1),
            "feh": (0.0, 0.1),
            "parallax": (116.0, 1.0),
            "maxAV": 0.05
        },
        "prot": 2.2,        # rotation period in days
        "prot_err": 0.2
    },
    {
        "name": "HD 209458",
        "params": {
            "teff": (6117, 50),
            "logg": (4.38, 0.05),
            "feh": (0.02, 0.05),
            "parallax": (21.24, 0.1),
            "maxAV": 0.1
        },
        "prot": 14.4,       # rotation period in days
        "prot_err": 1.0
    }
]
```

```python
# AP Col and HD 209458 test without rotation
for s in stars:
    print(f"\nRunning {s['name']}...")

    star = sd.Star(s["params"])   # no rotation period
    star.fit()                    # run default of 100,000

    age, errp, errm, _ = star.age_results()

    print(f"{s['name']} Age = {age:.2f} +{errp:.2f} -{errm:.2f} Gyr")
```

```python
# AP Col and HD 209458 test with rotation
for s in stars:
    print(f"\nRunning {s['name']} with rotation...")

    star = sd.Star(
        s["params"],
        prot=s["prot"],
        prot_err=s["prot_err"]
    )

    star.fit()   # default ~100,000 samples
    age, errp, errm, _ = star.age_results()

    print(f"{s['name']} Age = {age:.2f} +{errp:.2f} -{errm:.2f} Gyr")
```
